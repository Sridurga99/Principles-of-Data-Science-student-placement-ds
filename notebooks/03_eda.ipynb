```python
plt.figure(figsize=(12, 8))

correlation_matrix = df.select_dtypes(include="number").corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

```python
from scipy.stats import chi2_contingency

categorical_features = ["gender", "degree", "branch"]

for feature in categorical_features:
    table = pd.crosstab(df[feature], df["placed"])
    chi2, p_value, dof, expected = chi2_contingency(table)

    print(f"\n{feature.upper()}")
    print("Chi-square statistic:", round(chi2, 4))
    print("p-value:", round(p_value, 6))

    if p_value < 0.05:
        print("Result: Statistically significant association with placement")
    else:
        print("Result: No statistically significant association with placement")
```

```python
from scipy.stats import mannwhitneyu

numerical_features = [
    "cgpa",
    "backlogs",
    "internships",
    "certifications",
    "coding_skills",
    "communication_skills",
    "aptitude_score",
    "projects"
]

for feature in numerical_features:
    not_placed = df.loc[df["placed"] == 0, feature]
    placed = df.loc[df["placed"] == 1, feature]

    statistic, p_value = mannwhitneyu(
        not_placed,
        placed,
        alternative="two-sided"
    )

    print(f"{feature}:")
    print("U statistic:", round(statistic, 2))
    print("p-value:", round(p_value, 6))

    if p_value < 0.05:
        print("Result: Statistically significant difference\n")
    else:
        print("Result: No statistically significant difference\n")
```
